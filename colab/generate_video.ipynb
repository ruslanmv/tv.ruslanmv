## 2️⃣ Configure API Keys

```python
import os
from google.colab import userdata

# Optional: Configure API keys from Colab Secrets
# Go to Tools > Secrets to add these keys

try:
    # YouTube credentials
    os.environ['YOUTUBE_CLIENT_ID'] = userdata.get('YOUTUBE_CLIENT_ID')
    os.environ['YOUTUBE_CLIENT_SECRET'] = userdata.get('YOUTUBE_CLIENT_SECRET')
    os.environ['YOUTUBE_REFRESH_TOKEN'] = userdata.get('YOUTUBE_REFRESH_TOKEN')
    print('✅ YouTube credentials configured')
except:
    print('⚠️  YouTube credentials not found in Colab Secrets')

try:
    # Cloudflare R2 credentials (optional)
    os.environ['R2_ACCOUNT_ID'] = userdata.get('R2_ACCOUNT_ID')
    os.environ['R2_ACCESS_KEY_ID'] = userdata.get('R2_ACCESS_KEY_ID')
    os.environ['R2_SECRET_ACCESS_KEY'] = userdata.get('R2_SECRET_ACCESS_KEY')
    print('✅ R2 credentials configured')
except:
    print('ℹ️  R2 credentials not found (optional)')

try:
    # Optional: TTS API key
    os.environ['ELEVENLABS_API_KEY'] = userdata.get('ELEVENLABS_API_KEY')
    print('✅ ElevenLabs API key configured')
except:
    print('ℹ️  Using free TTS (gTTS)')
```

```python
# Show results
import json

with open('output/upload_results.json', 'r') as f:
    results = json.load(f)

print('\n📊 Upload Results:')
print(json.dumps(results, indent=2))

# Show YouTube URL
if 'youtube' in results['platforms'] and results['platforms']['youtube']['status'] == 'success':
    print(f"\n📺 YouTube: {results['platforms']['youtube']['url']}")

# Show R2 URL
if 'r2' in results['platforms'] and results['platforms']['r2']['status'] == 'success':
    print(f"💾 R2 Backup: {results['platforms']['r2']['url']}")
```
